```python
# Do a random walk in dimension d, and report how long it took to return.
def one_random_walk(dimension,verbose=False):

    startposition = dimension*[0]
    position = dimension*[0]
    numsteps = 0
    maxdist = 0
    maxposition = dimension*[0]

    if verbose:
        route = [tuple(position)]

    while(True):
        #print(f"Currently at {position}")
        whichaxis = ZZ.random_element(0,2)
        direction = ZZ.random_element(0,2)
        numsteps += 1

        if direction == 0:
            position[whichaxis] += 1
        else:
            position[whichaxis] += -1

        curdist = sum([abs(xx) for xx in position])
        if curdist > maxdist:
            maxdist = curdist
            maxposition = tuple(position)

        if verbose:
            route.append(tuple(position))

        if position == startposition: # If back, return the answer.
            if not verbose:
                return numsteps
            else:
                return [numsteps,maxdist,maxposition,route]
```

```python
# This cell will just run a few random walks and tell you the result.
for j in range(2):
    [myreturn,mymax,mymaxpos,myroute] = one_random_walk(2,verbose=True)
    print(f"Returned in {myreturn} steps. Far point was {mymaxpos}.")
    if(myreturn < 50):
        print(f"Route: {myroute}")
    else:
        print("Route: too long to print")
    print()
```
